## Part 4

1. Total sales revenue of the company

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

2. Top-selling electronic device

```python
top_product = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print("Top-selling product:")
print(top_product.head(1))
```

3. Total revenue by electronic device

```python
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

print(product_revenue)
```

4. Category Analysis

Find: Revenue generated by each category

```python
# Task 4 — Category Analysis

# Calculate total revenue for each category
category_revenue = df.groupby("Category")["Revenue"].sum()

# Sort from highest to lowest
category_revenue = category_revenue.sort_values(ascending=False)

# Display revenue generated by each category
print("Revenue Generated by Each Category:")
print(category_revenue)
```

5. Monthly Sales Analysis

Find: Monthly revenue generated by the company

```python
# Task 5 — Monthly Sales Analysis

# Calculate total revenue for each month
monthly_revenue = df.groupby(df["Order_Date"].dt.month)["Revenue"].sum()

# Sort from January to December
monthly_revenue = monthly_revenue.sort_index()

# Display monthly revenue
print("Monthly Revenue:")
print(monthly_revenue)
```

## Part 5 — Data Visualization

Bar Chart - Revenue by City

```python
# Bar Chart - Revenue by City

city_revenue = df.groupby("City")["Revenue"].sum()

city_revenue.plot(
    kind="bar",
    figsize=(10, 5),
    title="Revenue by City"
)

plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

2. Pie Chart - Category Contribution

```python
# Pie Chart - Category Contribution

category_revenue = df.groupby("Category")["Revenue"].sum()

category_revenue.plot(
    kind="pie",
    figsize=(7, 7),
    autopct="%1.1f%%",
    title="Category Contribution"
)

plt.ylabel("")
plt.show()
```

3. Line Chart — Monthly Sales Trend

```python
# Line Chart - Monthly Sales Trend

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

monthly_revenue = df.groupby(
    df["Order_Date"].dt.to_period("M")
)["Revenue"].sum()

monthly_revenue.index = monthly_revenue.index.astype(str)

monthly_revenue.plot(
    kind="line",
    figsize=(12, 5),
    marker="o",
    title="Monthly Sales Trend"
)

plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
